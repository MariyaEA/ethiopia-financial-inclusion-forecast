Event-augmented trend model with pessimistic (65%), base (100%) and optimistic (135%) event realization.

```
from pathlib import Path
import sys, pandas as pd, numpy as np, matplotlib.pyplot as plt
ROOT=Path.cwd().parent if Path.cwd().name=="notebooks" else Path.cwd(); sys.path.insert(0,str(ROOT))
from src.data_loader import load_unified_data,split_records,validate_schema
df=load_unified_data(ROOT/"data/raw/ethiopia_fi_unified_data.csv"); R=split_records(df); observations=R["observation"]; events=R["event"]; impact_links=R["impact_link"]; targets=R["target"]
from src.forecasting import all_scenarios
f=all_scenarios(observations,impact_links); display(f.round(2))
```

```
for code,g in f.groupby("indicator_code"):
 for s,z in g.groupby("scenario"): plt.plot(z.year,z.forecast,marker="o",label=s)
 b=g[g.scenario.eq("Base")]; plt.fill_between(b.year,b.lower,b.upper,alpha=.2); plt.title(code); plt.legend(); plt.show()
```

Usage has greater event-driven upside than access. Key uncertainties are policy execution, activity conversion, rural reach, affordability, inflation, connectivity and source-definition changes.
